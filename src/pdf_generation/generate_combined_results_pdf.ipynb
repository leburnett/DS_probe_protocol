```python
pdf_specs = {
    "pdf_w": 216
  , "pdf_h": 280
  , "pdf_res": 192
  , "font_size": 15
  , "pdf_margin": [5, 5]
}
```

Possible 'stim_type's = 'bar2', 'bar6', 'edge', 'grating', 'flash'

```python
for stim_type in ['bar2', 'edge']:
    
    for idx, row in log_table.iterrows():

        plot_info = {
            "stim_type": stim_type
            , "cell_type": row['cell_type']
            , "date_str": row['date_str']
            , "date_to_process": row['date_to_process']
        }
        generate_combined_pdf(plot_info=plot_info, pdf_specs=pdf_specs)
```
